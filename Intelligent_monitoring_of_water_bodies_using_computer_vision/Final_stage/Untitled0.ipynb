```python
from ultralytics import YOLO

# Загружаем предобученную модель
model = YOLO('yolov8n.pt')

# Запускаем обучение
results = model.train(
    data='/content/dataset/data.yaml',   # Локальный путь к файлу в Colab
    epochs=50,
    imgsz=640,
    project='/content/drive/MyDrive/Архипелаг_2026/runs', # Результаты сохранятся прямо на Гугл Диск
    name='my_experiment',

    verbose=True,      # Полный и детальный вывод всех метрик в консоль
    save=True        # Сохранение контрольных точек (чекпоинтов) модели
)
```
